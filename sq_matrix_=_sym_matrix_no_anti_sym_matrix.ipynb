```python
''' we know in square matrix can be decomposes to symetric matrix and antisymetric matrix.
there only contribution is from symmetric matrix, anti symmetric matrix part vanishes.
this is numerical proof.
In multivariate gaussian there is covariance matrix in the exponent.
In this covariance matrix or inverse of covariance matrix is called precision matrix
the antisymmetric part doesnot contribute to the exponent to the gaussian distribution only contribution from symetric part.
Since the inverse of a symmetric matrix is also symmetric, the covariance matrix Σ can also be chosen symmetric.
This is why both covariance and precision matrices are always taken as symmetric.'''

import numpy as np

# Define a random 2D precision matrix (not necessarily symmetric)
Lambda = np.array([[2, 3],
                   [-1, 4]])

# Decompose into symmetric (S) and antisymmetric (A) parts
S = 0.5 * (Lambda + Lambda.T)
A = 0.5 * (Lambda - Lambda.T)

# Pick a random vector y = (x - mu)
y = np.array([1.5, -2.0])

# Compute quadratic forms
Q_full = y.T @ Lambda @ y
Q_sym = y.T @ S @ y
Q_antisym = y.T @ A @ y

Q_full, Q_sym, Q_antisym
```

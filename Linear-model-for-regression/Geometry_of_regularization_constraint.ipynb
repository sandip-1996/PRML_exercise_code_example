```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------
# 1. Create grid for w1, w2
# --------------------------
w1 = np.linspace(-3, 3, 200)
w2 = np.linspace(-3, 3, 200)
W1, W2 = np.meshgrid(w1, w2)

# --------------------------
# 2. Define error function
# (center shifted to mimic PRML figure)
# --------------------------
# Minimum at (1, 1)
E = (W1 - 1)**2 + (W2 - 1)**2

# --------------------------
# 3. Plot function
# --------------------------
def plot_constraint(ax, constraint_type="L2", eta=3):
    # Blue contours (error surface)
    ax.contour(W1, W2, E, levels=5)

    # L2 constraint: circle
    if constraint_type == "L2":
        theta = np.linspace(0, 2*np.pi, 200)
        x = np.sqrt(eta) * np.cos(theta)
        y = np.sqrt(eta) * np.sin(theta)
        ax.plot(x, y)

    # L1 constraint: diamond
    elif constraint_type == "L1":
        x = np.linspace(-eta, eta, 200)
        y1 = eta - np.abs(x)
        y2 = -y1
        ax.plot(x, y1)
        ax.plot(x, y2)

    # axes
    ax.axhline(0)
    ax.axvline(0)
    ax.set_aspect('equal')
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)

# --------------------------
# 4. Plot both cases
# --------------------------
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

plot_constraint(axes[0], "L2")
axes[0].set_title("L2 (Ridge)")

plot_constraint(axes[1], "L1")
axes[1].set_title("L1 (Lasso)")

plt.show()
```

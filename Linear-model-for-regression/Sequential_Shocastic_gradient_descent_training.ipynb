```python
# shocastic gradient descent
''' Sequential learning of least square error'''
```

```python
import numpy as np
import matplotlib.pyplot as plt

N = 100
# generate synthetic data
X = np.linspace(0,10,N) # 100 data between o to 10
```

This is gradient desent algorithm for the case of sum of square error function

```python
true_w0 = 2.0
true_w1 = 1.5
T = true_w0 + true_w1*X + np.random.normal(0,1,N)
'''
here true_w0 + true_w1*X = y(x,w) and normal distrbution is noise
this is equation 3.8 see notebook'''

# design matrix with basis function phi(x)
phi = np.vstack([np.ones(N), X]).T

# let weight is 0
w = np.zeros(2)

# consider learning rate .01
eta = 0.01

# SGD training
epochs = 20
loss_history = []

for epoch in range(epochs):
  indices = np.random.permutation(N)
  # it suffles the indices upto 0 to N-1

  for n in indices:
    phi_n = phi[n]
    t_n = T[n]

    y_n = np.dot(w,phi_n)

    error = t_n - y_n

   #SGD update
    w = w + eta*error*phi_n

  y_all = phi@w
  loss = 0.5*np.mean((T - y_all)**2)
  loss_history.append(loss)

print("learned weights:", w)

# plot the data
plt.scatter(X, T, label = "data")
plt.plot(X, y_all, color = 'red', label = 'SGD fit')
plt.legend()
plt.title("SGD Linear Regression")
plt.show()

# plot the loss curve
plt.plot(loss_history)
plt.title("Loss vs Epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

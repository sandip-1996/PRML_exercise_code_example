```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt
```

```python
'''
bionomial distribution, find pmf
'''
n = 50 # number of coin flips
p = 0.6 # probability of getting head is 0.6 biased.
k = 30 # what is the probability of getting exactly 30 success.
probability_of_k = binom.pmf(k,n,p)
print(f'the probability of getting {k} success in {n} flips:', probability_of_k)
```

```python
''' find cdf cumulative distribution function '''
# what is the probability of getting less than k success.
CDF_k = binom.cdf(k,n,p)
print(f'the probability of getting less than {k} success in {n} flips:', CDF_k)
```

```python
''' generate random success
here we generate number of success after 100 coin flips for 10 samples '''
random_samples = binom.rvs(n,p,size = 10)
probability_random_samples = binom.pmf(random_samples,n,p)
print(random_samples)
print(f'probability of success of random_samples:', probability_random_samples)
```

```python
# visualize this random success after getting 50 trials
plt.figure(figsize=(5,3))
plt.bar(random_samples, probability_random_samples, width = 0.5, color = 'red',label = 'PMF')
plt.title(f'bionomial distribution of 10 samples of {n} trails each, {p} probability')
plt.xlabel(f'number of success, {random_samples}')
plt.ylabel('probability')
plt.xticks(random_samples)
plt.grid(axis = 'y',linestyle = '--',alpha = 0.9)
plt.legend()
plt.show()
```

```python
# iterate range 50, find pmf of each success
x_values = np.arange(0,51)
print(x_values)
probability_x_values = binom.pmf(x_values,n,p)
print(probability_x_values)
```

```python
# visualize the data
plt.figure(figsize=(8,4))
plt.bar(x_values,probability_x_values,color = 'red',label = 'PMF')
plt.title(f'bionomial distribution of {n} trails, {p} probability')
plt.xlabel('success (heads)')
plt.ylabel('probability')
# plt.xticks(x_values)
plt.grid(axis='y', linestyle='--', alpha=1.0)
plt.legend()
plt.show()
```

```python
# plot the CDF
''' we have 'x_values' as success we will find CDF according to each success'''
CDF_x_values = binom.cdf(x_values,n,p)
print(CDF_x_values)
```

```python
plt.figure(figsize = (10,4))
plt.bar(x_values, CDF_x_values, color = 'red', label = 'PMF')
plt.xlabel(f'probability of getting less than k success in {n} flips')
plt.ylabel('probability')
plt.title(f'cumulative distribution function')
plt.xticks()
plt.legend()
plt.show()
```
